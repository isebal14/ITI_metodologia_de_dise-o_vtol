# VTOL de detección de incendios
TEMA: Metodología de diseño de un dron VTOL para la detección de incendios forestales.  

Profesor patrocinante: Bernardo Hernandez 

Descripción: Los incendios forestales son una problemática que afecta a nuestro país habiendo 28 hasta hoy este año 2026 https://www.conaf.cl/incendios/situacion-actual-y-pronostico-de-incendios/, donde las herramientas de detección implementadas en la acutalidad son estáticas o menos eficientes energéticamente que un aeronave de ala fija. Por lo que para la detección y actuación rápida de incendios forestales se propone una metodolgía de diseño de una aeronave autónoma de despegue y aterrizaje vertical para monitoreo de zonas con riesgo de incenidos forestales.
